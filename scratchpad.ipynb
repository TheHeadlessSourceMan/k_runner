```python
import typing
import os
import k_runner

processes=k_runner.findSystemProcesses('Logseq')
connections:typing.List["pconn"]=[]
for process in processes:
    for conn in process.net_connections():
        if conn.family==2:
            connections.append(f'{conn.laddr.ip}:{conn.laddr.port}')
print(connections)
```
